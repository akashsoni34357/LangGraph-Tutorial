```python
from langgraph.graph import StateGraph,START,END
from langchain_openai import ChatOpenAI
from typing import TypedDict
from dotenv import load_dotenv
```

```python
load_dotenv()

model=ChatOpenAI()
```

```python
#define a state

class BlogState(TypedDict):

    title:str
    outline:str
    content:str
    eval_score:str
```

```python
def create_outline(state:BlogState)-> BlogState:

    #fetch title
    title=state['title']

    #call llm and generate outline
    res_outline=model.invoke(f'Generate the outline for blog on the title {title}').content

    #update the state
    state['outline']=res_outline

    return state
```

```python
def create_blog(state:BlogState)->BlogState:

    #fetch title and current outline
    title=state['title']
    outline=state['outline']


    #call llm ang generate the detailed blog on outline

    prompt=f'Generate the detailed blog on the title -{title} using the following outline - {outline}'
    res_content=model.invoke(prompt).content

    #update the state
    state['content']=res_content

    return state
```

```python
def evaluation_score(state:BlogState)->BlogState:

  #fetch the outline and blog 
  outline=state['outline']
  content=state['content']

  #call the llm and generate the score out of 10
  prompt=f'based on the given outline-{outline} ,evaluate the following content-{content} on the score of 10 '
  eval_score=model.invoke(prompt).content

  #update the state
  state['eval_score']=eval_score

  return state
```

```python
graph= StateGraph(BlogState)

#create a node

graph.add_node('create_outline',create_outline)
graph.add_node('create_blog',create_blog)
graph.add_node('evaluation_score',evaluation_score)

#create edges
graph.add_edge(START,'create_outline')
graph.add_edge('create_outline','create_blog')
graph.add_edge('create_blog','evaluation_score')
graph.add_edge('evaluation_score',END)


#compile the graph

workflow=graph.compile()


```

```python
#execute 

initial_state={'title':'Rise of corruption in indian politics ','eval_score':''}

final_state=workflow.invoke(initial_state)

print(final_state)
```
